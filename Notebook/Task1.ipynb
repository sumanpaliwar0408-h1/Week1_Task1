# Task 1 — E-Commerce Dataset Cleaning

## Objective

The objective of this task is to inspect, clean, validate, and prepare an e-commerce customer shopping behavior dataset for reliable analysis.

The cleaning process focuses on identifying missing values, checking duplicate records, standardizing column names, validating important fields, removing redundant information, and creating useful derived features while preserving the original raw dataset.

## Dataset Overview

The dataset contains customer shopping behavior records, including customer demographics, purchased items, product categories, purchase amounts, locations, review ratings, subscription status, shipping methods, discount usage, previous purchases, payment methods, and purchase frequency.

The raw dataset contains **3,900 rows and 18 columns**. Each row represents a customer shopping record identified by a `Customer ID`.

The raw dataset is retained separately and all cleaning operations are performed on a loaded copy in the notebook.

```python
# Loading Dataset
import pandas as pd

df = pd.read_csv("../Data/Raw/customer_shopping_behavior.csv")
```

## 1. Initial Data Inspection

The dataset was first inspected using `head()`, `info()`, and `describe()` to understand its structure, data types, dimensions, value ranges, and categorical distributions.

The raw dataset contains **3,900 observations and 18 features**. The features include numerical variables such as age, purchase amount, review rating, and previous purchases, as well as categorical variables describing customer demographics and shopping behavior.

Initial inspection also showed that most columns were complete, while the `Review Rating` column contained a small number of missing values. Further checks were therefore performed before making any cleaning decisions.

```python
# Checking for missing data or null values
df.isnull().sum()
```

```python
missing_report = pd.DataFrame({
    'missing_count': df.isnull().sum(),
    'missing_percentage': (df.isnull().mean() * 100).round(2)
})

missing_report[missing_report['missing_count'] > 0]
```

## 2. Missing Value Analysis

A column-wise missing-value check was performed using both the number and percentage of missing observations.

Only the `Review Rating` column contained missing data:

- Missing values: **37**
- Missing percentage: **0.95%**
- Total records: **3,900**

Since the missing proportion was very small and the affected records still contained useful customer and purchase information, removing the rows would unnecessarily discard valid data. Therefore, imputation was preferred over deletion.

```python
# Imputing missing values in Review Rating column with the median rating of the product category (each category)
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

### Missing-Value Treatment

Missing `Review Rating` values were filled using the **median review rating within the corresponding product category**.

A category-level median was selected instead of a single overall dataset value so that the imputed rating reflects the typical rating pattern of the relevant product category. The median is also less sensitive to extreme values than the mean.

After imputation, the dataset was checked again and **no missing values remained**.

```python
df.isnull().sum()
```

```python
# Snake_casing for uniformity

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

## 3. Column Name Standardization

Column names were standardized to lowercase `snake_case` format to improve readability and consistency when working with the dataset programmatically.

For example:
- `Customer ID` → `customer_id`
- `Item Purchased` → `item_purchased`
- `Review Rating` → `review_rating`
- `Frequency of Purchases` → `frequency_of_purchases`
`Purchase Amount (USD)` was simplified to `purchase_amount`.
#This produces consistent and easier-to-reference feature names throughout the remaining workflow.

```python
df.columns
```

## 4. Feature Engineering
Two additional features were created to make the cleaned dataset more useful for subsequent customer analysis-
### Age Group

The numerical `age` variable was divided into four quantile-based customer groups:
- Young Adult
- Adult
- Middle-aged
- Senior
This converts continuous age values into broader customer segments that can be used for demographic comparison while retaining the original `age` column.

```python
# New column- age_group
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# New column- purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

### Purchase Frequency in Days

The categorical `frequency_of_purchases` variable was mapped to an approximate numerical number of days between purchases.

Examples include:
- Weekly → 7 days
- Fortnightly / Bi-Weekly → 14 days
- Monthly → 30 days
- Quarterly / Every 3 Months → 90 days
- Annually → 365 days
The original categorical column was retained, while `purchase_frequency_days` provides a numerical representation that may be more convenient for later analysis.

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

## 5. Data Validation
Validation checks were performed on important numerical and categorical fields to identify logically invalid or inconsistent values.

## Numerical Validation
The following conditions were checked:
- `age` must be greater than 0.
- `purchase_amount` must be greater than 0.
- `previous_purchases` must not be negative.

The validation produced:
- Invalid age values: **0**
- Invalid purchase amounts: **0**
- Invalid previous-purchase values: **0**
Therefore, no records required removal or correction based on these numerical validation rules.

```python
numeric_validation = {
    'invalid_age': (df['age'] <= 0).sum(),
    'invalid_purchase_amount': (df['purchase_amount'] <= 0).sum(),
    'invalid_previous_purchases': (df['previous_purchases'] < 0).sum()
}

pd.Series(numeric_validation)
```

```python
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    whitespace_issues = (
        df[col].notna() &
        (df[col] != df[col].str.strip())
    ).sum()

    print(f"{col}: {whitespace_issues} whitespace issues")
```

### Categorical Consistency Check

Categorical columns were checked for leading or trailing whitespace that could create artificial duplicate categories during analysis.
No whitespace inconsistencies were detected in the categorical fields. Therefore, no additional whitespace cleaning was required.

```python
df[['discount_applied','promo_code_used']].head(10)
```

## 6. Redundant Feature Check

The `discount_applied` and `promo_code_used` columns appeared to contain identical information. Their values were compared row by row to determine whether one of the columns was redundant.
The comparison confirmed that the two columns matched for every record in the dataset. Keeping both would therefore duplicate the same information without adding analytical value.
As a result, `promo_code_used` was removed and `discount_applied` was retained.

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
# Dropping promo code used column

df = df.drop('promo_code_used', axis=1)
```

The result showed:
**Fully duplicated rows: 0**
Therefore, no records were removed as duplicates. This follows the principle of removing records only when duplication is confirmed rather than deleting potentially valid observations.
